```python
def linear_searching(list1,target):
    for i in range(len(list1)):
        if list1[i]==target:

            return f"index of the element is : {i}"
    return -1
l=[20,39,50,70]
e=23
linear_searching(l,e)
```

```python
def binary_search(list1,target):
    l=sorted(list1)
    start=0
    last=len(l)-1
    while(start<=last):
        mid=(start+last)//2
        if(l[mid]==target):
            return(f"element found at {mid} index")
            
        elif l[mid]<target:
            start=mid+1
        else:
            last=mid-1
    print("not found")
    return -1
list1 = [1, 2, 3, 4, 10, 9, 100, 50]
target = 10
print(binary_search(list1,target))

        
```
