# Searching

## 21.
Write a program to perform linear search on a list.


```python
def linear_search(list1,target):
    for i in range(len(list1)):
        if list1[i] ==target:
            return(f"{target} found at {i}")
    else:
        return("Not found")
print(linear_search([1,23,4,5,60,10],0))
```
