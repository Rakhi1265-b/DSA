```python
def binarey_search(list1,target):
    list1=sorted(list1)
    start=0
    end=len(list1)-1
    while(start<=end):
        mid=(start+end)//2
        if(list1[mid]==target):
            return(f"found at {mid}")
        elif(list1[mid]<target):
            start=mid+1
        elif(list1[mid]>target):
            end=mid-1
    
    return "no"
print(binarey_search([1,23,4,5,60,10],10))
    

```

# Bubble Sorting

```python
a=[10,20,12,18,30,26,25]
for i in range(len(a)-1):
    for j in range(len(a)-i-1):
        if(a[j]>a[j+1]):
            a[j],a[j+1]=a[j+1],a[j]
print(a)
```

# Selection Sorting

```python
a=[10,20,12,18,30,26,25]
for i in range(len(a)):
    min=i
    for j in range(i+1,len(a)):
        if(a[min]>a[j]):
            min=j
    a[i],a[min]=a[min],a[i]
            
print(a)
```

```python
def primes(m):
    p=2
    c=0
    while(True):
        for k in range(2,p):
            if p%k==0:
                break
        else:
            c+=1
            if c==m:
                return p
        p+=1
n=int(input())
for i in range(1,n+1):
    l=i
    for j in range(1,i+1):
        print(primes(l),end=" ")
        l-=1
    print()
    
```
